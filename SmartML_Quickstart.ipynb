## 1. Load Data (1 Line)

```python
# Enable auto-reload so the notebook always uses the latest installed module
%load_ext autoreload
%autoreload 2

import smartml as sml

# Load dataset (supports CSV, Excel, JSON, Parquet)
df = sml.load(r"f:\smartml_assistant_module\archive\AI_Usage_and_Impact_on_Students_and_Professionals.csv")
df.head(3)
```

## 2. One-Line Supervised AutoML (Classification / Regression)
Cleans data, selects best algorithm across 9 candidate models, computes feature importance, plots charts, and produces HTML reports.

```python
# 1 Line to train and benchmark all models
result = sml.auto(df, target="Would_Recommend")

# Show leaderboard and summary
print(result.summary)
```

## 3. Semi-Supervised, Reinforcement & Self-Supervised Learning (1 Line Each)

```python
# 1. Semi-Supervised Learning (Pseudo-labeling on unlabeled target rows)
semi_res = sml.semi_supervised(df, target="Would_Recommend")
print("--- Semi-Supervised ---")
print(semi_res.narrative)

# 2. Reinforcement Learning / Contextual Bandits (Offline Decision Optimization)
rl_res = sml.reinforcement(df, action="AI_Tool", reward="Satisfaction_Score")
print("\n--- Reinforcement Learning (Bandits) ---")
print(rl_res.narrative)
print("Arm pull counts:", rl_res.arm_pull_counts)

# 3. Self-Supervised Representation Learning (Dense Latent Embeddings)
ssl_res = sml.self_supervised(df, target="Would_Recommend")
print("\n--- Self-Supervised Autoencoder ---")
print(ssl_res.narrative)
print("Latent embeddings shape:", ssl_res.embeddings_df.shape)
```

## 4. Unsupervised Clustering & Time-Series Forecasting (1 Line Each)

```python
# 1. Unsupervised Clustering (isolates target to prevent leakage)
cluster_res = sml.clustering(df, n_clusters=3)
print(cluster_res.summary)

# 2. Multi-Step Time-Series / Sequence Forecasting with Lag Features
fc_res = sml.forecast(df, target="Productivity_Score", horizon=5)
print(fc_res.narrative)
print("Next 5 steps predictions:", [round(p, 2) for p in fc_res.predictions])
```
